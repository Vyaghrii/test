```python
import os
import osmnx as ox
```

```python
# Define place name and output directory
place_name = "Delhi, India"
output_dir = r"C:\Users\ASUS\Desktop\CODING Stuff\test"
```

```python
# Create folder if it doesn't exist
os.makedirs(output_dir, exist_ok=True)
```

```python
# Download the driveable road network from OpenStreetMap
print(f"Downloading road network for {place_name}...")
G = ox.graph_from_place(place_name, network_type="drive")
```

```python
# Convert the graph to GeoDataFrames (nodes and edges)
nodes, edges = ox.graph_to_gdfs(G)
```

```python
# Export shapefiles
print("Saving shapefiles...")
nodes.to_file(os.path.join(output_dir, "nodes.shp"))
edges.to_file(os.path.join(output_dir, "edges.shp"))
```

```python
print("✅ Export completed successfully!")
print(f"Files saved in: {output_dir}")
```
